#### Primeira consulta no arquivo para entender as primeiras informações das músicas mais populares

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df_high = pd.read_csv('high_popularity_spotify_data.csv', encoding='utf-8', sep=';')
print(df_high.head())
df_high.info()
df_high.describe()
```

#### Primeira consulta no arquivo para entender as primeiras informações das músicas menos populares

```python
df_low = pd.read_csv('low_popularity_spotify_data.csv', encoding='utf-8', sep=';')
print(df_low.head())
df_low.info()
df_low.describe()
```

### União dos dados

```python
df_high["classe_popularidade"] = "high"
df_low["classe_popularidade"] = "low"
df = pd.concat([df_high, df_low], ignore_index=True)
print(df.info())
```

#### Colunas numéricas principais

```python
colunas_numericas = [
    "energy", "danceability", "valence", "tempo",
    "loudness", "acousticness", "speechiness",
    "instrumentalness", "track_popularity"
]

df[colunas_numericas] = df[colunas_numericas].apply(pd.to_numeric, errors="coerce")
```

### Média por grupo (high x low)

```python
df.groupby("classe_popularidade")[colunas_numericas].mean()
```

```python
comparacao_media = (
    df.groupby("classe_popularidade")[colunas_numericas]
    .mean()
    .T
)

comparacao_media["diff"] = comparacao_media["high"] - comparacao_media["low"]
comparacao_media.sort_values("diff", ascending=False)
```

### Popularidade média por gênero

```python
df.groupby("playlist_genre")["track_popularity"].mean().sort_values(ascending=False)
```

### Gênero mais comum

```python
df["playlist_genre"].value_counts()
```

### Comparação de energia por classe de popularidade

```python
sns.boxplot(
    data=df,
    x="classe_popularidade",
    y="energy"
)
plt.title("Distribuição de Energy por Classe de Popularidade")
plt.xlabel("Classe de Popularidade")
plt.ylabel("Energy")
plt.show()
```

### Comparação de danceability por classe de popularidade

```python
sns.boxplot(
    data=df,
    x="classe_popularidade",
    y="danceability"
)
plt.title("Distribuição de Danceability por Classe de Popularidade")
plt.xlabel("Classe de Popularidade")
plt.ylabel("Danceability")
plt.show()
```

### Relação acousticness com popularidade

```python
sns.boxplot(
    data=df,
    x="classe_popularidade",
    y="acousticness"
)
plt.title("Distribuição de Acousticness por Classe de Popularidade")
plt.xlabel("Classe de Popularidade")
plt.ylabel("Acousticness")
plt.show()
```

### Distribuição de danceability por classe de popularidade

```python
sns.histplot(
    data=df,
    x="danceability",
    hue="classe_popularidade",
    bins=30,
    kde=True,
    element="step"
)
plt.title("Distribuição de Danceability por Classe de Popularidade")
plt.show()
```

### Energia X Popularidade

```python
sns.scatterplot(
    data=df,
    x="energy",
    y="track_popularity",
    hue="popularity_class",
    alpha=0.6
)
plt.title("Energy vs Popularidade da Faixa")
plt.show()

```

```python
sns.barplot(
    data=df,
    x="popularity_class",
    y="danceability",
    estimator="mean"
)
plt.title("Danceability Média por Classe de Popularidade")
plt.show()
```

```python
sns.barplot(
    data=df,
    x="playlist_genre",
    y="track_popularity",
    estimator="mean"
)
plt.xticks(rotation=45)
plt.title("Popularidade Média por Gênero")
plt.show()
```

### Conversão de ms para min

```python
df["duration_min"] = df["duration_ms"] / 60000
```

### Gráfico de outliers da duração das músicas por popularidade

```python
sns.boxplot(
    data=df,
    x="classe_popularidade",
    y="duration_min"
)
plt.title("Duração das Faixas (em minutos) por Classe de Popularidade")
plt.ylabel("Duração (min)")
plt.show()

```
